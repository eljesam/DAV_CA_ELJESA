## 3. Basic cells: variables, functions, control flow
Demonstrate Python variables, functions, classes, loops, and displaying multiline outputs.

```python
# Variables
project_name = 'Data Analysis Notebook'
version = 1.0

# Function definition

def greet(name):
    return f"Hello, {name}! Welcome to the notebook."

# Class definition
class DataSample:
    def __init__(self, values):
        self.values = values
    
    def mean(self):
        return np.mean(self.values)

# Control flow and output
numbers = list(range(1, 6))
for n in numbers:
    print(f"Number: {n}")

print(greet('Analyst'))
print('Sample mean:', DataSample(numbers).mean())
```

## 4. Load and inspect data
Load datasets from CSV or JSON, display head(), info(), describe(), dtypes, and a sample of rows.

```python
data_folder = os.path.join('..', 'Data')
csv_path = os.path.join(data_folder, 'edri_data.csv')
json_path = os.path.join(data_folder, 'sample_data.json')

if os.path.exists(csv_path):
    df = pd.read_csv(csv_path)
    print('Loaded CSV:', csv_path)
elif os.path.exists(json_path):
    df = pd.read_json(json_path)
    print('Loaded JSON:', json_path)
else:
    df = pd.DataFrame({'example_col': [1, 2, 3]})
    print('Using example DataFrame because no CSV/JSON file was found.')

print('Shape:', df.shape)
print('Columns:', df.columns.tolist())

display(df.head())
print('\nInfo:')
df.info()
print('\nDescribe:')
display(df.describe(include='all'))
print('\nData types:')
print(df.dtypes)
print('\nSample rows:')
display(df.sample(min(5, len(df))))
```

## 5. Data cleaning & preprocessing
Handle missing values, convert types, filter rows, and create tidy DataFrame pipelines.

```python
# Example preprocessing pipeline
if 'example_col' in df.columns:
    df_clean = (
        df
        .copy()
        .assign(example_col=lambda x: pd.to_numeric(x['example_col'], errors='coerce'))
        .dropna(subset=['example_col'])
        .reset_index(drop=True)
    )
else:
    df_clean = df.copy()

# Show changes and missing values
print('Missing values per column:')
print(df_clean.isna().sum())

display(df_clean.head())
```

## 6. Exploratory visualizations (Matplotlib / Seaborn)
Create line, scatter, histogram, bar, and heatmap visualizations with labels and figure sizing.

```python
plt.figure(figsize=(10, 4))
if 'example_col' in df_clean.columns:
    sns.histplot(df_clean['example_col'], kde=True)
    plt.title('Example Column Distribution')
    plt.xlabel('example_col')
    plt.ylabel('Count')
    plt.show()

# Example scatter plot using synthetic data if no numeric columns exist
x = np.linspace(0, 10, 50)
y = np.sin(x)
plt.figure(figsize=(8, 4))
plt.plot(x, y, marker='o')
plt.title('Example Line Plot')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.grid(True)
plt.show()
```

## 7. Interactive widgets and inputs (ipywidgets)
Use interactive widgets to update parameters, plots, and display callbacks inside the notebook.

```python
slider = widgets.IntSlider(value=5, min=1, max=20, description='Points')

def update_plot(points=5):
    x = np.linspace(0, 2 * np.pi, points)
    y = np.sin(x)
    plt.figure(figsize=(7, 3))
    plt.plot(x, y, marker='o')
    plt.title(f'Sine wave with {points} points')
    plt.show()

widgets.interact(update_plot, points=slider)
```

## 9. Debugging cells and using VS Code notebook features
Use VS Code notebooks with the integrated debugger, cell output pane, breakpoints, and variable inspection.

```python
def compute_stats(values):
    total = sum(values)
    count = len(values)
    average = total / count
    breakpoint()  # Set a breakpoint in VS Code notebook for inspection
    return total, count, average

sample_values = [1, 2, 3, 4, 5]
compute_stats(sample_values)
```

## 10. Unit tests from notebook / integrated test runner
Add pytest examples and run tests via the terminal or VS Code test explorer.

```python
# Example inline tests using assertions

def test_compute_stats():
    total, count, average = compute_stats([2, 4, 6])
    assert total == 12
    assert count == 3
    assert average == 4

print('Inline test run:')
test_compute_stats()
print('All inline assertions passed.')

# To run pytest from the terminal:
# !pytest -q
```
